```python
from google.colab import userdata
GROQ_API = userdata.get('groq_key')
```

```python
import os
os.environ['GROQ_API'] = GROQ_API
```

```python
import aisuite as ai
client = ai.Client()

models = ["groq_key:llama3-8b-8192"]

messages = [
    {"role": "system", "content": "give me summery."},
    {"role": "user", "content": "AI."},
]

for model in models:
    response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=0.75
    )
    print(response.choices[0].message.content)
```
