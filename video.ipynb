```python
from google import genai
from google.genai import Client

client : Client = genai.Client(
    api_key= GEMINI_API_KEY,

)
model: str = "gemini-2.0-flash-exp"
```

```python
video_file_name = "/content/quaid video.mp4"
```

```python
from google.genai.types import Content, Part

# Define the video URI and MIME type
video_uri = "https://generativelanguage.googleapis.com/v1beta2/files/example_video_uri"  # Replace with the actual URI
mime_type = "video/mp4"  # Specify the correct MIME type

# Define the model and prompt
model = "your_model_name"  # Replace with your model's name
prompt = """
For each scene in this video, generate captions that describe the scene
along with any spoken text placed in quotation marks. Place each caption
into an object with the time code of the caption in the video.
"""

# Generate content
response = client.models.generate_content(
    model=model,
    contents=[
        Content(
            role="user",
            parts=[
                Part.from_uri(uri=video_uri)  # Use Part.from_uri to handle video input
            ]
        )
    ],
    prompt=prompt
)

# Print the generated response
print(response.text)
```

```python
import time

def upload_video(video_file_name):
    video_file = client.files.upload(path=video_file_name)
    while video_file.state == "processing ":
        print('waiting for video to be processed')
        time.sleep(10)
        video_file = client.files.get(name=video_file.name or"")

    if video_file.state == "FAILED":
       raise ValueError(video_file.state)
    print(f"'Video processing complete:" + (video_file.uri or ""))
    return video_file
my_video = upload_video(video_file_name)
```
